```python
# 1. Import the sqlite3 module (comes with python)
import sqlite3

# 2. Connect to a database file named "CollegeDB"
#    If the file does not exist, SQLite creates it.
conn = sqlite.connect("CollegeDB")

# 3. Create a "Cursor" object.
#    The cursor is like a handle that leds you send SQL commands to the database.
cur = conn.cursor()

# 4. Create a table called "CollegeDB" if it does not already exists.
#    This SQL command defines the structure (columns) of the table.
cur.execute("""
CREATE TABLE IF NOT EXISTS college (
   roll_no INTEGER PRIMARY KEY,    --unique ID for each student
   name TEXT NOT NULL,             --students's name (cannot be copied)
   branch TEXT,                    --e.g., "BCA", "BBA"
   semester INTEGER                --current semester number
   )
   """)
 # 5. Prepare some sample data as a list of tuples.
 #    Easch tuple represents one row to insert.
 data = [
     (1, "Arun", "BCA", 1),
     (2, "Priya", "BBA", 1),
     (3, "Rahul", "BCA", 2),
 ]

 # 6. Insert multiple rows into the "CollegeDB" table.
 #    The SQL uses placeholders (?) to safely pass values from python.
 cur.executemany(
     "INSERT OR REPLACE INTO students (roll_no, branch, semester) VALUES (?, ?, ?, ?)",
     data,
 )

# 7. Commit the transaction.
#    This saves all changes (inserts/updates/deletes) to the database file.
conn.commit()

# 8. Run a select query to read data.
#    This SQL asks for roll_no, name, and branch of students in semester 1.
cur.execute("SELECT roll no. name. branch FROM students WHERE semester = ?". (1.))

# 9. Fetch all matching rows from the result.
#    fetchall() returns a list of tuples, one per row.
rows = cur.fetchall()

# 10. Loop through the rows and print them.
for row in rows:
    print(row)

# 11. Update: increase semester by 1 for all BCA students.
cur.execute("UPDATE students SET semester = semester + 1 WHERE branch = ?" , ("BCA",))
conn.commit()

#12. Delete: remove the student with roll_no = 3.
cur.execute("DELETE FROM students WHERE roll_no = ?", (3))
conn.commit()

#13. Close the connection when done.
#    This releases the database file and cleans up the resources.
conn.close()
(1, 'Arun', 'BCA')
(2, 'Priya', 'BBA')
```

```python
# analyze_students.py
import sqlite3
import pandas as pd

# -------------------------
# PART A: Connect to the existing database (INSTRUCTOR PREPARED)
# -------------------------

# Connect to the same "students.db" that was created by create_students_db.py
conn = sqlite3.connect("students.db")

# Load the "students" table into a pandas DataFrame.
# From here on, students work only with df_students (like an Excel sheet).
df_students = pd.read_sql_query("SELECT * FROM students", conn)

```
